```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import VotingClassifier

from sklearn.model_selection import train_test_split
from sklearn.datasets import make_moons

from sklearn.metrics import accuracy_score

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## Ensemble Learning
    Using the performance of multiple models instead of settling with a single model. These multiple models can either be from the same or different algorithms

#### Voting
    Make multiple models (different algorithms) and perform voting on the results to decide the final classification

```python
X, y = make_moons(n_samples=500, noise=0.30, random_state=42)
moon_data = pd.DataFrame(X,columns=['Feature_1','Feature_2'])
moon_data['Label'] = y
moon_data
```

```python
X_train, X_test, y_train, y_test = train_test_split(moon_data[['Feature_1','Feature_2']],moon_data['Label'],random_state=42)
```

```python
sns.scatterplot(X_train['Feature_1'],X_train['Feature_2'],hue=moon_data['Label'],legend=False)
sns.scatterplot(X_test['Feature_1'],X_test['Feature_2'],hue=moon_data['Label'],edgecolor='black',legend=False)
```

```python
log_reg = LogisticRegression()
sv_classifier = SVC()
rfc = RandomForestClassifier()
voting_clf = VotingClassifier([('lr',log_reg),('svm',sv_classifier),('rf',rfc)],voting='hard')
```

```python
for clf in (log_reg,sv_classifier,rfc,voting_clf):
    clf.fit(X_train,y_train)
    pred = clf.predict(X_test)
    print(clf.__class__.__name__,accuracy_score(y_test,pred))
```

    The hard voting uses maximum votes based on the output of all algorithms

```python
log_reg = LogisticRegression()
sv_classifier = SVC(probability=True)
rfc = RandomForestClassifier()
voting_clf = VotingClassifier([('lr',log_reg),('svm',sv_classifier),('rf',rfc)],voting='soft')
```

```python
for clf in (log_reg,sv_classifier,rfc,voting_clf):
    clf.fit(X_train,y_train)
    pred = clf.predict(X_test)
    print(clf.__class__.__name__,accuracy_score(y_test,pred))
```

    Instead of directly considering the output, soft voting uses the predicted class probabilities of each class then average the probabilities for each class and output the class with max probability. For this reason, soft voting performs better than it's hard counterpart often

#### Bagging - Bootstrap Aggregation
    Using the same algorithm but on different random samples of data. This sampling is done with replacement as in any given data point can occur in one predictor more than once.

```python
from sklearn.ensemble import BaggingClassifier
```

```python
bag_cls = BaggingClassifier(base_estimator=SVC(probability=True),n_estimators=10,bootstrap=True,n_jobs=-1,max_samples=100)
bag_cls.fit(X_train,y_train)
pred = bag_cls.predict(X_test)
accuracy_score(y_test,pred)
```

```python
sv_classifier = SVC(probability=True)
sv_classifier.fit(X_train,y_train)
pred = sv_classifier.predict(X_test)
accuracy_score(y_test,pred)
```

    We can see a clear distinction with and without bagging. Bagging by it's nature has almost similar bias error but reduces the variance factor and hence the overall improvement in performance 

#### Pasting
    Using the same algorithm on different random samples without replacement as in any given data point can't occur more than once in a predictor

```python
bag_cls = BaggingClassifier(base_estimator=SVC(probability=True),bootstrap=False,max_samples=100)
bag_cls.fit(X_train,y_train)
pred = bag_cls.predict(X_test)
accuracy_score(y_test,pred)
```

```python
sv_classifier = SVC(probability=True)
sv_classifier.fit(X_train,y_train)
pred = sv_classifier.predict(X_test)
accuracy_score(y_test,pred)
```

    Because bootstraping includes diversity in the predictors, the results are slightly better than that of pasting

*The results for Bagging and pasting usually have bias around that of the individual predictors in the ensemble but have a lower variance*

#### Out of the bag evaluation(oob)
    In bagging there can be a repetition of any given data point in a predictor, when the number of data points in each bootstrapped sample is same as the data points in train data, there's usually 1-(1/e) unique entries in each bootstrapped sample. This is 63% of the data, the rest of 37% can be used as the test set to measure this predictor's performance even before we get to testing.
    More often than not, performance on the oob is close to that of the test set

```python
bag_cls = BaggingClassifier(base_estimator=SVC(),bootstrap=True,oob_score=True,n_jobs=-1,n_estimators=50)
bag_cls.fit(X_train,y_train)
bag_cls.oob_score_
```

```python
pred = bag_cls.predict(X_test)
accuracy_score(y_test,pred)
```

    The oob method works only if we are working with bootstrapped sampling

#### Random Subspace
    When it is possible to use sub samples of data, it can also be possible to use sub samples of features. So if we use a sub-sample to train predictors without bootstrapping records and use all training data in samples, then it is random subspace

#### Random Patch
    If we include both bootstrapping and random subspace then it's a random patch

#### Boosting
    In it's literal sense boosting is improving on what is existing, it is nothing different in machine learning as well. While building ensembles, we make an initial predictor and build up on it to correct the existing errors so as to improve performance. Two most frequently used boosting techniques are:
    - Adaboost
    - XGboost

***Adaboost***, short for Adaptive boosting as the name suggests adapts to work on learning the previously wrongly classified instances better. The output is determined by a weighted average or voting of the individual classifiers of the ensemble. The weight of each classifier depends on it's accuracy

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import AdaBoostRegressor
from sklearn.datasets import load_boston
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

bs = load_boston()
target = bs['target']
bs = pd.DataFrame(data = bs['data'],columns = bs['feature_names'])
bs['target'] = target

X_train, X_test, y_train, y_test = train_test_split(bs.drop(labels=['target'],axis=1),bs['target'],test_size=0.2)
```

```python
abr = AdaBoostRegressor(base_estimator = LinearRegression(),n_estimators=100)
abr.fit(X_train,y_train)
```

```python
pred = abr.predict(X_test)
print(np.sqrt(mean_squared_error(y_test,pred)))
```

***XGboost*** short for extreme Gradient boost builds a classifier, then another on the residuals of the previous one, on and on...

```python
import xgboost
from xgboost import XGBRegressor
```

```python
bs = load_boston()
bs['feature_names']
```

```python
target = bs['target']
bs = pd.DataFrame(data = bs['data'],columns = bs['feature_names'])
bs['target'] = target
bs.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(bs.drop(labels=['target'],axis=1),bs['target'],test_size=0.2,random_state=3)
```

```python
eval_set = (X_test,y_test)
xgbr = XGBRegressor()
xgbr.fit(X = X_train,y = y_train,eval_set = [eval_set],early_stopping_rounds = 10)
```

```python
dtrain = xgboost.DMatrix(X_train,y_train)
dtest = xgboost.DMatrix(X_test,y_test)
params = {'objective':'reg:linear'}
xgboost.train(params=params,dtrain=dtrain,early_stopping_rounds=10,evals=[(dtest,'test'),(dtrain,'train')])
```
